```python
dbutils.widgets.text("endpoint", "chains")
endpoint = dbutils.widgets.get("endpoint")
```

```python
import os
import requests
import pandas as pd
from datetime import datetime

api_key = dbutils.secrets.get("cijene-api", "api-key")
headers = {"Authorization": f"Bearer {api_key}"}

response = requests.get(f"https://api.cijene.dev/v1/{endpoint}/", headers=headers)
response.raise_for_status()
data = response.json()

df = pd.DataFrame(data[endpoint])
display(df)

folder = f"/Volumes/workspace/cijene/landing/{endpoint}"
os.makedirs(folder, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
df.to_csv(f"{folder}/{endpoint}_{timestamp}.csv", index=False)
```
